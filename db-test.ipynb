```python
import numpy as np
import duckdb
import pandas as pd
```

```python
con = duckdb.connect()

# Set up a base view pointing to your parquet file
print(con.execute("""
    CREATE VIEW master_table AS 
    SELECT * FROM 'data/MasterTable-2026_09_02.parquet';
"""))
```

```python
startDate = pd.to_datetime('2025-1-1')
endDate = pd.to_datetime('2025-12-1')
con = duckdb.connect(database=':memory:')
startDateStr = startDate.strftime("%Y-%m-%d")
endDateStr = endDate.strftime("%Y-%m-%d")


query = f"""
SELECT *
FROM "data/MasterTable-2026_09_02.parquet"
WHERE ServiceType = 'Housing' 
  AND (
    ( StartDate >= '{startDateStr}'::TIMESTAMP_NS AND EndDate <= '{endDateStr}'::TIMESTAMP_NS ) 
    OR ( EndDate >= '{startDateStr}'::TIMESTAMP_NS AND EndDate <= '{endDateStr}'::TIMESTAMP_NS )
    OR ( StartDate >= '{startDateStr}'::TIMESTAMP_NS AND StartDate <= '{endDateStr}'::TIMESTAMP_NS )
    );
"""



tbl = con.execute(query).fetch_df()
con.close()
```
